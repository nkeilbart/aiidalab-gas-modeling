```python
import ipywidgets as widgets
from IPython.display import display

# File picker (only works for local files in JupyterLab or compatible environments)
file_picker = widgets.FileUpload(
    accept='',  # Accept all file types
    multiple=False  # Single file only
)

# Text box
text_box = widgets.Text(
    value='',
    placeholder='Type something',
    description='Input:',
    disabled=False
)

# Dropdown menu
dropdown = widgets.Dropdown(
    options=['Option 1', 'Option 2', 'Option 3'],
    value='Option 1',
    description='Choose:',
    disabled=False,
)

# Output area
output = widgets.Output()

# Button to process inputs
button = widgets.Button(description="Submit")

def on_button_clicked(b):
    with output:
        output.clear_output()
        print("Text entered:", text_box.value)
        print("Dropdown selection:", dropdown.value)
        if file_picker.value:
            file_info = list(file_picker.value.values())[0]
            print("File uploaded:", file_info['metadata']['name'])
        else:
            print("No file uploaded.")

button.on_click(on_button_clicked)

# Display the widgets
display(widgets.VBox([file_picker, text_box, dropdown, button, output]))
```
